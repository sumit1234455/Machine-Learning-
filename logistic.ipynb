```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
```

```python
data = load_breast_cancer()
X = data.data
y = data.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
def sigmoid(z):
  z=np.clip(z, -500, 500)
  return 1/(1+np.exp(-z))
```

```python
def trained(X,y,learning_rate=0.01,epochs=1000):
  m,n=X.shape
  w=np.zeros(n)
  b=0
  for i in range(epochs):
    z=np.dot(X,w)+b
    y_pred=sigmoid(z)

    dw = (1/m) * np.dot(X.T, (y_pred - y))
    db = (1/m) * np.sum(y_pred - y)
    w -= learning_rate * dw
    b -= learning_rate* db
  return w,b
```

```python
def predict(X,w,b):
  z=np.dot(X,w)+b
  y_pred=sigmoid(z)
  return (y_pred>=0.5).astype(int)
```

```python
w,b=trained(X_train,y_train)
y_pred=predict(X_test,w,b)
```
