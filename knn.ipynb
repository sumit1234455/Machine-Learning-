```python
import math
```

```python
def distance(a,b):
  return math.sqrt(sum((a[i] - b[i])**2 for i in range(len(a))))
```

```python
def knn(data,label,test_p,k):
  dist=[]
  for i in range(len(data)):
    d=distance(data[i],test_p)
    dist.append((d,label[i]))
  dist.sort(key=lambda x: x[0])
  count = {}
  for i in range(k):
        label = dist[i][1]
        count[label] = count.get(label, 0) + 1

  return max(count, key=count.get)
```

```python
def f1_score(actual, predicted):
    TP=FP=FN=0
    for i in range(len(actual)):
        if actual[i] == 1 and predicted[i] == 1:
            TP += 1
        elif actual[i] == 0 and predicted[i] == 1:
            FP += 1
        elif actual[i] == 1 and predicted[i] == 0:
            FN += 1
    if TP == 0:
        return 0
    precision= TP / (TP + FP) if (TP + FP) > 0 else 0
    recall=TP/ (TP + FN) if (TP + FN) > 0 else 0
    if precision+recall == 0:
        return 0

    return 2 * (precision * recall) / (precision + recall)
```

```python
def find_best_k(data, labels):
    n = len(data)
    max_k = int(math.sqrt(n))
    best_k = 1
    best_score = -1
    for k in range(1, max_k + 1):
        predictions = []
        for point in data:
            pred = knn(data, labels, point, k)
            predictions.append(pred)

        score = f1_score(labels, predictions)
        print(f"K = {k}, F1 Score = {score:.2f}")

        if score > best_score:
            best_score = score
            best_k = k
    return best_k, best_score

data = [
    [1, 2], [2, 3], [3, 3],
    [6, 5], [7, 7], [8, 6],
    [1, 0], [0, 1], [9, 8], [8, 9]
]

labels = [0,0,1,1,1,1,0,0,1,1]
best_k, best_score = find_best_k(data, labels)
print("\nBest K:", best_k)
print("Best F1 Score:", round(best_score, 2))
```

```python
import math
import random

def distance(a, b):
    return math.sqrt(sum((a[i] - b[i])**2 for i in range(len(a))))

def knn(train_data, train_labels, test_point, k):
    dist = []

    for i in range(len(train_data)):
        d = distance(train_data[i], test_point)
        dist.append((d, train_labels[i]))

    dist.sort(key=lambda x: x[0])

    count = {}
    for i in range(k):
        label = dist[i][1]
        count[label] = count.get(label, 0) + 1

    return max(count, key=count.get)

def f1_score(actual, predicted):
    TP = FP = FN = 0
    for i in range(len(actual)):
        if actual[i] == 1 and predicted[i] == 1:
            TP += 1
        elif actual[i] == 0 and predicted[i] == 1:
            FP += 1
        elif actual[i] == 1 and predicted[i] == 0:
            FN += 1

    if TP == 0:
        return 0

    precision = TP / (TP + FP) if (TP + FP) > 0 else 0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0

    if precision + recall == 0:
        return 0

    return 2 * (precision * recall) / (precision + recall)

def cross_validation_f1(data, labels, k_neighbors, folds=5):
    combined = list(zip(data, labels))
    random.shuffle(combined)
    fold_size = len(data) // folds
    scores = []
    for i in range(folds):
        test = combined[i*fold_size:(i+1)*fold_size]
        train = combined[:i*fold_size] + combined[(i+1)*fold_size:]

        train_data = [x[0] for x in train]
        train_labels = [x[1] for x in train]
        test_data = [x[0] for x in test]
        test_labels = [x[1] for x in test]
        predictions = []
        for point in test_data:
            pred = knn(train_data, train_labels, point, k_neighbors)
            predictions.append(pred)

        score = f1_score(test_labels, predictions)
        scores.append(score)

    return sum(scores) / len(scores)

def find_best_k_cv(data, labels, max_k=10, folds=5):
    best_k = 1
    best_score = -1
    for k in range(1, max_k + 1):
        score = cross_validation_f1(data, labels, k, folds)
        print(f"K = {k}, Avg F1 Score = {score:.3f}")
        if score > best_score:
            best_score = score
            best_k = k

    return best_k, best_score
data = [
    [1, 2], [2, 3], [3, 3],
    [6, 5], [7, 7], [8, 6],
    [1, 0], [0, 1], [9, 8], [8, 9]
]
labels = [0,0,1,1,1,1,0,0,1,1]
best_k, best_score = find_best_k_cv(data, labels, max_k=6, folds=5)

print("\nBest K:", best_k)
print("Best F1 Score:", round(best_score, 3))
```
