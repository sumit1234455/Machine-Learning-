```python
import numpy as np
from sklearn import datasets
import math
```

```python
data = datasets.load_iris()
X = data.data
y = data.target
feature_names = data.feature_names
```

```python
class Node:
    def __init__(self, feature=None, threshold=None, left=None, right=None, value=None):
        self.feature = feature
        self.threshold = threshold
        self.left = left
        self.right = right
        self.value = value

```

```python
def count_labels(y):
    count = {}
    for value in y:
        if value not in count:
            count[value] = 0
        count[value] += 1
    return count

```

```python
def entropy(y):
  counts=count_labels(y)
  total=len(y)
  entr=0
  for i in counts:
    p=counts[i]/total
    entr+=-p*math.log(p,2)
  return entr
```

```python
def split(X,y,feature,threshold):
  left_x,left_y=[],[]
  right_x,right_y=[],[]
  for i in range((len(X))):
    if X[i][feature]<threshold:
      left_x.append(X[i])
      left_y.append(y[i])
    else:
      right_x.append(X[i])
      right_y.append(y[i])
  return left_x,left_y,right_x,right_y
```

```python
def informationgain(y,left_y,right_y):
  parent=entropy(y)
  n=len(y)
  nl=len(left_y)
  nr=len(right_y)
  child = (nl/n)*entropy(left_y) + (nr/n)*entropy(right_y)

  return parent-child


```

```python
def bestsplit(X,y):
  best_gain=-1
  best_feature=None
  best_threshold=None
  num_feature=len(X[0])
  for feature in range(num_feature):
    values=[]
    for row in X:
      if row[feature] not in values:
        values.append(row[feature])
    for threshold in values:
      left_x,left_y,right_x,right_y=split(X,y,feature,threshold)
      gain=informationgain(y,left_y,right_y)
      if gain>best_gain:
        best_gain=gain
        best_feature=feature
        best_threshold=threshold
  return best_feature,best_threshold
```

```python
def majority(y):
    counts = count_labels(y)
    max_count = -1
    label = None
    for i in counts:
        if counts[i] > max_count:
            max_count = counts[i]
            label = i
    return label
```

```python
def tree(X,y ,depth=0,maxdepth=3):
  if len(set(y))==1:
    return Node(value=y[0])
  if depth>=maxdepth:
    return Node(value=majority(y))

  feature,threshold=bestsplit(X,y)
  if feature is None:
    return Node(value=majority(y))
  left_X, left_y, right_X, right_y = split(X, y, feature, threshold)
  left = tree(left_X, left_y, depth+1, maxdepth)
  right =tree(right_X, right_y, depth+1, maxdepth)

  return Node(feature, threshold, left, right)
```

```python
def print_tree(node, depth=0):
    indent = "  " * depth


    if node.value is not None:
        print(indent + "Leaf:", node.value)
        return
    print(indent + f"{feature_names[node.feature]} <= {node.threshold}")
    print(indent + "-> Left:")
    print_tree(node.left, depth + 1)
    print(indent + "-> Right:")
    print_tree(node.right, depth + 1)

tree1 = tree(X, y, maxdepth=3)

print_tree(tree1)
```
